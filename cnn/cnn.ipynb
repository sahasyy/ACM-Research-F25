```python
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

transform = transforms.ToTensor()

# get traing daataa
train_set = datasets.MNIST(root='./data', train=True, download=True, transform=transform)
train_loader = DataLoader(train_set, batch_size=64, shuffle=True)

# get testing data
test_set = datasets.MNIST(root='./data', train=False, download=True, transform=transform)
test_loader = DataLoader(test_set, batch_size=64, shuffle=True)

train_set
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F


class CNN(nn.Module):
    def __init__(self):
        super().__init__()

        # one input channel to to 16
        self.conv1 = nn.Conv2d(in_channels=1, out_channels=16, kernel_size=3, padding=0, stride=1)
        # 16 to 32
        self.conv2 = nn.Conv2d(in_channels=16, out_channels=32, kernel_size=3, padding=0, stride=1)
        # input = channels * height * width
            # 28x28 input for MNIST
            # first conv is kernal size 3, stride 1, padading 0. so new dimension is (28 - 3 + 1) = 26
            # maxpool w/ stride 2 aand kernel 2, so output is 26/2 = 13
            # second conv is kernal size 3, stride 1, padading 0. so new dimension is (13 - 3 + 1) = 11
            # pooling: 11/2 = 5 for int division
            # final height and width = 5
            # 32 * 5 * 5 = 800

        # output = # of classes (10 for digits)
        self.fc = nn.Linear(in_features=32*5*5, out_features=10)
    
    def forward(self, x):
        x = self.conv1(x)
        x = F.relu(x)
        x = F.max_pool2d(input=x, kernel_size=2, stride=2)

        x = self.conv2(x)
        x = F.relu(x)
        x = F.max_pool2d(input=x, kernel_size=2, stride=2)

        # flatten, ( x.size(0) keeps the batch size, -1 flattens the rest)
        x = x.view(x.size(0), -1)
        # fully connected
        x = self.fc(x)
        # output
        return x
    
model = CNN()
```

```python
# training
import torch.optim as optim

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)
epochs = 10

for epoch in range(epochs):
    model.train()

    train_loss = 0.0

    for i, data in enumerate(train_loader, 0):
        # data = (inputs, labels)
            # inputs: batch of images (a tensor with shape [batch_size, channels, height, width])
            # labels: batch of corresponding labels (a tensor with shape [batch_size])
        inputs, labels = data

        # reset graadients
        optimizer.zero_grad()

        # forard pass
        output = model(inputs)

        # calc loss
        loss = criterion(output, labels)

        # backwaards pass
        loss.backward()

        # optimize based on gradients
        optimizer.step()  # update weights

        train_loss += loss.item()  # accumulate loss for monitoring

    # test accuracy on test set
    model.eval()

    total_predictions = 0
    total_correct = 0

    with torch.no_grad():
        for i, data in enumerate(test_loader, 0):
            inputs, labels = data

            # forward pass for predictions
            output = model(inputs)

            # get class labels
            predictions= torch.argmax(output, dim=1)

            total_correct += (predictions== labels).sum().item()
            total_predictions += labels.size(0)
    accuracy = total_correct / total_predictions
    print(f"epoch {epoch} had loss {train_loss} and test accuracy of {accuracy}")

        



```
